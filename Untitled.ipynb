```python






import os
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader

# 1. Create Directories
os.makedirs('dataset/train/cats', exist_ok=True)
os.makedirs('dataset/train/dogs', exist_ok=True)
os.makedirs('dataset/test/cats', exist_ok=True)
os.makedirs('dataset/test/dogs', exist_ok=True)
print("Folders created successfully.")

# 2. Device Setup
device = torch.device("cuda" if torch.cuda.is_available() else 

"cpu")

# 3. Data Transformations
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# 4. Load Data

train_dataset = datasets.ImageFolder(root='dataset/train', transform=transform)
test_dataset = datasets.ImageFolder(root='dataset/test', transform=transform)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)


# 5. Model Setup
model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 2)
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters

(), lr=0.0001)

# 6. Training
epochs = 5
print("Starting Training...")
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        
        optimizer.zero_grad()

        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item() * inputs.size(0)
        
    epoch_loss = running_loss / len(train_dataset) if len(train_dataset) > 0 else 0
    print(f"Epoch {epoch+1}/{epochs} - Loss: {epoch_loss:.4f}")


# 7. Testing
model.eval()
correct = 0
total = 0
print("Starting Testing...")
with torch.no_grad():
    for inputs, labels in test_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        outputs = model(inputs)
        _, preds = torch.max(outputs, 1)

        
        total += labels.size(0)
        correct += (preds == labels).sum().item()

if total > 0:
    accuracy = 100 * correct / total
    print(f"Test Accuracy: {accuracy:.2f}%")
else:
    print("Please add images to dataset/test folder.")
```
